### List comprehensions

Let's imagine we want a list of the squares all the even numbers in a range

```python
list_of_squares = []
for i in range(10):
    if i%2 == 0:
        list_of_squares.append(i**2)
list_of_squares
```

```python
list_of_squares = [i**2 for i in range(10) if i%2 == 0]
list_of_squares
```

## Functions

```python
def make_list(length, exponent):
    list_of_exponents = [i**exponent for i in range(length) if i%2 == 0]
    return list_of_exponents
```

```python
make_list(20, 3)
```

### Multiple conditions

Let's get a subset of data where cumulative cases fall within 1 million and 2 million:

Note that instead of using the `and` and `or` operands we use `&` and `|` respectively, which are the boolean operands in Pandas. Let's see other boolean conditions:

If we want any data of a specific list of WHO regions:

```python
df[df['WHO_region'].isin(['EUR', 'AMR', 'WPR'])].head()
```

```python
df[~df['WHO_region'].isin(['EUR', 'AMR', 'WPR'])].head()
```

We want to know the total number (globally) of vaccines administered per demographic group by the 1st quarter of 2024. Let's start filtering the data we need. Remember we're only interested in four columns: `YEAR`, `QUARTER`, `COVID_VACCINE_ADM_1D`, and `GROUP`.

# Data visualization

We will now use the data we've been seeing to create some plots. We will rely on the matplotlib library:

```python
import matplotlib.pyplot as plt

%matplotlib inline
```

Let's use the vaccination totals we obtained before to make our first plot:

```python
data = subset.groupby('GROUP').agg({'COVID_VACCINE_ADM_1D': 'sum', 'YEAR': 'max', 'QUARTER': 'max'})
```

Pyplot's `bar()` function takes two arguments to plot a bar chart: an array of categories for the X axis and an array of values for the Y axis. In this case we will use the DataFrame's indices for the X axis categories, and the COVID_VACCINE_ADM_1D totals for the Y axis values.

```python
plt.bar(data.index, data.COVID_VACCINE_ADM_1D)
plt.show()
```

This already looks good, but we can further improve it; for instance, we can add axis labels and a title:

```python
plt.bar(data.index, data.COVID_VACCINE_ADM_1D)

plt.ylabel('Total administered vaccines')
plt.xlabel('Demographic groups')
plt.title('Distribution of vaccines administered \n by demographic group')

plt.show()
```

A great thing about using Python for visualization is that we can use iterative operations to generate a large number of charts with little effort. Let's say we want to see the timeline of covid cases per WHO region. 

```python
# Let's start by getting a list of the unique region names:
regions = df['WHO_region'].unique()
regions
```

```python
# Now let's create a loop that iterates over each region to plot total cases of each region over time:
for region in regions: 
    data = df[df['WHO_region']==region].groupby('Date_reported').agg('sum') # We aggregate by date
    plt.plot(data.index, data.Cumulative_cases) # We use the plot function

plt.show()
```

We can see that the timelines are plotted but we don't know which line belongs to which region. Let's add some arguments:

```python
for region in regions: 
    data = df[df['WHO_region']==region].groupby('Date_reported').agg('sum')
    plt.plot(data.index, data.Cumulative_cases, label=region) # the label argument lets us give a name to the plotted line
    plt.legend() # The legend function will display the chart's legend
    plt.xlabel('Time')
    plt.ylabel('Cases')
    plt.title('Timeline of cases per region')

plt.show()
```

Now, let's imagine we want to have individual charts for each region. Pyplot can create a grid to host various charts:

```python
fig, axes = plt.subplots(nrows = 2, ncols = 4, figsize = (20, 5))
```

Let's have a look at the previous statement: the `subplots` function takes number of rows and number of columns argument, as well as the size of each figure. Since we have 7 WHO regions, we will create a 2x4 grid. The `subplots` function returns two objects: `fig` which is the figures to host the plots, and `axes` which is an array of figure identifiers:

```python
fig
```

```python
axes
```

This might sound confusing right now, but it will make more sense with the exercise. Let's now create a list of tuples that we'll use as coordinates of our grid:

```python
plot_cols = list((l1, l2) for l1 in list(range(2)) for l2 in list(range(4)))
plot_cols
```

As you can see, each element of the list is a tuple with a pair of coodinates of 0 to 3 (x axis) and 0 to 1 (y axis). Let's now try to make these plots:

```python
fig, axes = plt.subplots(nrows = 2, ncols = 4, figsize = (20, 5))

# We use the zip operator to iterate simultaneously over two separate iterables.
# This way we can pair a region to a figure in each iteration:
for (region, plot_col) in zip(regions, plot_cols):
    data = df[df['WHO_region']==region].groupby('Date_reported').agg('sum')
    axes[plot_col].plot(data.index, data.Cumulative_cases)
    
plt.show()
```

Let's now add the other elements of the plots

```python
fig, axes = plt.subplots(nrows = 2, ncols = 4, figsize = (20, 5))

for (region, plot_col) in zip(regions, plot_cols):
    data = df[df['WHO_region']==region].groupby('Date_reported').agg('sum')
    axes[plot_col].plot(data.index, data.Cumulative_cases)
    #to add individual figure elements, we use the methods 'set_title()', 'set_ylabel()', 'set_ylabel()'
    axes[plot_col].set_title(region)
    axes[plot_col].set_xlabel('Time')
    axes[plot_col].set_ylabel('Number of cases or deaths')
    
plt.show()
```

It's not bad, but the titles and the axis labels are too crammed. Let's give them some space:

```python
fig, axes = plt.subplots(nrows = 2, ncols = 4, figsize = (20, 5))

for (region, plot_col) in zip(regions, plot_cols):
    data = df[df['WHO_region']==region].groupby('Date_reported').agg('sum')
    axes[plot_col].plot(data.index, data.Cumulative_cases)
    axes[plot_col].set_title(region)
    axes[plot_col].set_xlabel('Time')
    axes[plot_col].set_ylabel('Number of cases or deaths')
    
plt.subplots_adjust(hspace=0.5, wspace=0.5)
plt.show()
```
